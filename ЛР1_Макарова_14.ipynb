### 3. Проверка пропусков

Пропущенные значения могут исказить результаты анализа, поэтому данные проверяются на их наличие. 

```python
# Проверить данные на наличие пропусков и устранить их, если они есть (пропуски необходимо либо удалить, либо заменить каким-то значением).
print(df.isna().sum())

df['Income'] = df['Income'].fillna(df['Income'].median())
df['Kidhome'] = df['Kidhome'].fillna(df['Kidhome'].median())
df['Dt_Customer'] = df['Dt_Customer'].fillna(df['Dt_Customer'].mode()[0])
df['NumDealsPurchases'] = df['NumDealsPurchases'].fillna(df['NumDealsPurchases'].median())

print(df.isna().sum())
```


**Методы обработки пропусков:**
1. **Удаление строк** (`dropna()`) - применяется, если пропусков много или строка не несет какой-то смысл.
2. **Заполнение значениями** (`fillna()`) - применяется, если важно сохранить данные в других столбцах:
   - Для числовых данных: среднее арифметическое, медиана, мода,
   - Для категориальных данных: мода (наиболее частое значение) или константа ("Неизвестно").

**Выбранный подход:** Заполнение пропусков, так как их немного: 15 из 796 записей, и удаление строк привело бы к потере ценной информации по другим признакам.


---

**Результаты проверки:**

**До обработки:**
- Income: 12 пропусков

- Dt_Customer: 1 пропуск
- NumDealsPurchases: 1 пропуск

**Всего:** 15 пропущенных значений в 4 столбцах из 8

**После обработки:**

**Обоснование выбора методов:**

1. **Income → медиана:** Столбец содержит выбросы (максимальный доход 162,397 при минимальном 2,447). Медиана устойчива к выбросам и лучше отражает характерное значение, чем среднее арифметическое.

2. **Kidhome → медиана:** Дискретная переменная (0, 1, 2 ребёнка). Медиана = 0, что логично: у большинства клиентов нет детей.

3. **Dt_Customer → мода:** Дата регистрации - категориальная переменная в данном контексте. Мода (наиболее частая дата) сохраняет распределение клиентов по времени регистрации.

4. **NumDealsPurchases → медиана:** Дискретная переменная (количество покупок). Медиана = 2, что соответствует типичному поведению клиента.

**Альтернативные подходы:**
- Удаление строк с пропусками (dropna()) - потерялось бы 15 записей.
- Заполнение средним арифметическим - менее устойчиво к выбросам, особенно для Income.
- Заполнение константой (0 или -1) - могло бы исказить статистические показатели.

### 4. Проверка дубликатов

Дубликаты бывают двух видов - явные и неявные. Каждый из них может искажать результаты анализа данных, поэтому от них нужно избавляться на этапе предобработки.

#### Проверка явных дубликатов

**Явные дубликаты** - строки, которые полностью повторяют другие строки во всех столбцах. Обнаруживаются методом:

```python
print(df.duplicated().sum())
```

После обнаружения их нужно удалить с помощью метода drop_duplicates(). reset_index(drop=True) сбрасывает индексацию, чтобы не было пропусков в индексах.

```python
# удалите дубликаты, если они есть
df = df.drop_duplicates().reset_index(drop=True)
print(df.duplicated().sum())
```

После удаления снова проверяем что явных дубликатов больше не осталось.

#### Проверка неявных дубликатов


**Неявные дубликаты** - альтернативные написания одного и того же значения (разный регистр, опечатки, сокращения). Обнаруживаются методом unique() для категориальных столбцов, исправляются методом replace().


```python
print('Уникальные значения Education')
print(df['Education'].unique())
print()
print('Уникальные значения Marital_Status')
print(df['Marital_Status'].unique())
```

```python
# удалите дубликаты, если они есть
df['Marital_Status'] = df['Marital_Status'].replace('MARRIED', 'Married')
df['Marital_Status'] = df['Marital_Status'].replace('SINGL', 'Single')

print('Уникальные значения Education')
print(df['Education'].unique())
print()
print('Уникальные значения Marital_Status')
print(df['Marital_Status'].unique())
```

---
В результате проверки было обнаружено 4 явных дубликата. После удаления методом drop_duplicates() количество дубликатов стало 0.
Явные дубликаты представляют собой полные копии записей, которые не несут дополнительной информации. Их наличие искажает статистические показатели. 

Проверенные столбцы:
- Education,
- Marital_Status. 
Неявные дубликаты в столбце Education отсутствуют. Все значения записаны единообразно, без опечаток и разного регистра.

В столбце Marital_Status были обнаружены проблемы: MARRIED и Married - разный регистр у одного и того же значение, SINGL и Single - разный регистр и опечатка. После исправления регистр стал единым, опечатка была убрана. 

Значения Single и Alone были оставлены как отдельные категории, так как скорее всего они несут разный смысл:
- Single - юридический семейный статус (не в браке, холост/не замужем). Человек может жить с соседом, родителями или друзьями.
- Alone - факт проживания в одиночку. Человек может быть разведен (Divorced) или вдовцом (Widow), но жить один.
Их объединение могло приветси к потере ценной информации для анализа покупательского поведения и дохода.

- Строки - уровень образования (Education).

```python
# код выполнения задания
from datetime import datetime

# разница между текущей датой и датой регистрации клиента в компании
df['Years_since_registration'] = (datetime.now() - df['Dt_Customer']).dt.days / 365.25

print(df[['Dt_Customer', 'Years_since_registration']].head())
```

Категоризация клиентов по сроку регистрации.
Разделим клиентов на 3 категории:
- новый (от 10 до 13 лет).
- средний (от 13 до 15 лет).
- старый (более 15 лет).

Границы выбраны исходя из анализа данных, которые указывают на то, что все клиенты зарегестрировались более 12 лет назад.

```python
# столбец "Категория клиента"
bins_client = [10, 13, 15, 50] 
labels_client = ['новый', 'средний', 'старый']

df['Категория_клиента'] = pd.cut(df['Years_since_registration'], bins=bins_client, labels=labels_client)
print(df['Категория_клиента'].value_counts())
```

Категоризация по уровню дохода клиентов. Разделим годовой доход на 3 категории, опираясь на квантили, полученные ранее:
- низкий (до 40 000) - ниже 25-го перцентиля.
- средний (от 40 000 до 70 000) - между 25-м и 75-м перцентилями.
- высокий (выше 70 000) - выше 75-го перцентиля.

Использование статистических квантилей позволяет создать сбалансированные группы, избегая искажений из-за выбросов в доходах.

```python
# столбец "Категория дохода"
bins_income = [0, 40000, 70000, 200000] 
labels_income = ['низкий', 'средний', 'высокий']

df['Категория_дохода'] = pd.cut(df['Income'], bins=bins_income, labels=labels_income)
print(df['Категория_дохода'].value_counts())
```

**фильтрация данных** Нужно отфильтровать датасет по нескольким условиям одновременно:
1. Доход: средний и высокий.
2. Категория клиента: только новые.
3. Семейный статус: Топ-2 статуса по среднему сроку пребывания в компании.
4. Образование: Топ-3 типа образования по среднему доходу.

```python
top_marital = df.groupby('Marital_Status')['Years_since_registration'].mean().nlargest(2).index.tolist()
print(f"Топ-2 семейных статуса по сроку пребывания: {top_marital}")

top_education = df.groupby('Education')['Income'].mean().nlargest(3).index.tolist()
print(f"Топ-3 образования по среднему доходу: {top_education}")

filtered_df = df[
    (df['Категория_дохода'].isin(['средний', 'высокий'])) &
    (df['Категория_клиента'] == 'новый') &
    (df['Marital_Status'].isin(top_marital)) &
    (df['Education'].isin(top_education))
]

print(f"\nРазмер исходного датасета: {len(df)} строк.")
print(f"Размер отфильтрованного датасета: {len(filtered_df)} строк.")
```

**Создание сводных таблиц**

**1. Расчет возраста:**

Границы выбраны на основе описательной статистики датасета:
- низкий: до 40 000 (ниже 25-го перцентиля, ~36 141)
- средний: от 40 000 до 70 000 (между 25-м и 75-м перцентилями, ~69 293)
- высокий: выше 70 000 (выше 75-го перцентиля, максимум ~162 397)
Использование квантилей позволяет создать статистически обоснованные группы, устойчивые к выбросам.

**3. Фильтрация данных:**
- Доход: только средний и высокий (целевая платежеспособная аудитория).
- Возраст: старше 30 лет.
Этот порог выбран, чтобы отсечь аномально молодых клиентов и обеспечить достаточный размер выборки для репрезентативности.
- Семейный статус и Образование: отфильтрованы по топ-2 самым популярным категориям в датасете, чтобы анализировать наиболее массовые сегменты.

```python
import datetime

current_year = datetime.datetime.now().year
df['Age'] = current_year - df['Year_Birth']

bins_income = [0, 40000, 70000, 200000]
labels_income = ['низкий', 'средний', 'высокий']
df['Категория_дохода'] = pd.cut(df['Income'], bins=bins_income, labels=labels_income)

top_2_marital = df['Marital_Status'].value_counts().nlargest(2).index.tolist()
top_2_education = df['Education'].value_counts().nlargest(2).index.tolist()

print(f"Топ-2 семейных статуса: {top_2_marital}")
print(f"Топ-2 типа образования: {top_2_education}")

filtered_18 = df[
    (df['Категория_дохода'].isin(['средний', 'высокий'])) &
    (df['Age'] > 30) &
    (df['Marital_Status'].isin(top_2_marital)) &
    (df['Education'].isin(top_2_education))
]

print(f"\nРазмер исходной выборки: {len(df)}")
print(f"Размер отфильтрованной выборки: {len(filtered_18)}")

pivot_18 = filtered_18.pivot_table(
    index='Education',
    columns='Marital_Status',
    values='Income',
    aggfunc=['mean', 'median'],
    observed=True
).round(2)

print("\nСводная таблица")
display(pivot_18)
```

Отфильтрованная выборка из 274 клиентов старше 30 лет со средним/высоким доходом и самыми популярными статусами (Married, Together) и образованием (Graduation, PhD). Клиенты с образованием PhD имеют более высокий средний доход (66 837–67 904) по сравнению с Graduation (63 603–65 392). При этом медианные значения (~62–67 тыс.) ниже средних, что указывает на наличие клиентов с аномально высоким доходом. Статус Together демонстрирует чуть более высокие показатели дохода, чем Married, в обеих образовательных группах.
